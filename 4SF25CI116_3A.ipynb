```python
# We can observe that the entries of the first seven columns are shifted one postion to the left
# so we will shift the entries of these columns one position to the right.
df["Price"]=df["Required age"]
df["Required age"]=df["Peak CCU"]
df["Peak CCU"]=df["Estimated owners"]
df["Estimated owners"]=df["Release date"]
df["Release date"]=df["Name"]
df["Name"]=df["AppID"]
df["AppID"]=df.index + 1
```

```python
df.head()
```

```python
# The columns "Reviews", "Metacritic url", "Score Rank", "Movies", "Website", "Support url" and "Notes" have many missing values, so we will drop them from the dataset.
df=df.drop(columns=["Reviews","Metacritic url","Score rank","Movies","Website","Support url","Notes"])
```

```python
# Many games are free to play, so we will create a new column "Is Free" that will be True if the game is free and False if it is not.
df["Is Free"]=df["Price"]==0
```

```python
# Convert the datatype of the "Release date" column to datetime format.
df["Release date"]=pd.to_datetime(df["Release date"])
```

```python
df
```

```python
df["Release_year"]=df["Release date"].dt.year
```

```python
df["Release_year"]
```

```python
release_year=df["Release_year"].value_counts()
release_year.plot(kind="bar",color="blue")
```

```python
devices=df[["Windows","Mac","Linux"]].sum()
labels=["Windows","Mac","Linux"]
colors=["skyblue","brown","red"]
explode=(0.1, 0, 0)
plt.pie(devices,explode=explode,labels=labels,colors=colors,shadow=True,autopct='%1.1f%%')
plt.title('Game Platform Distribution')
plt.show()
```
